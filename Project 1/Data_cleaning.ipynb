```python
#SPEAK ONE LANGUAGE 
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

#  Les Valeurs fréquentes de CouponCode

```python
df['CouponCode'].value_counts()
```

```python
#remplacer les valeurs manquantes par No Coupon
df['CouponCode']=df['CouponCode'].fillna('No Coupon')
```

```python
# VERIFICATION APRE LE CHANGEMENT
df.isna().sum()
```

```python
#verifier le totalprice 
df["CalculatedTotal"] = (
    df["Quantity"] * df["UnitPrice"]
).round(2)

df["TotalPrice"] = df["TotalPrice"].round(2)

errors = df[
    df["TotalPrice"] != df["CalculatedTotal"]
]

print("Incorrect TotalPrice:", len(errors))
```

```python
# supprimer la colon CalculatedTotal avant saved the dataset
df.drop(columns=["CalculatedTotal"], inplace=True)
```

```python
df.to_excel(
    "Dataset_Cleaned.xlsx",
    index=False
)

print("Cleaned dataset saved successfully!")
```
